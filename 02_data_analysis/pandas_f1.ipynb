**Filtre resultados de **Ferrari** ou **Mercedes** com **pontos >= 15**. Atenção ao veneno textual.**

```python
resultados_df['equipe'] = resultados_df['equipe'].str.strip().str.capitalize()
```

```python
resultados_df['equipe'].value_counts()
```

```python
filter = ((resultados_df['equipe'] == 'Ferrari') | (resultados_df['equipe'] == 'Mercedes')) & (resultados_df['pontos'] >= 15)

resultados_df[filter].head(7)
```

**Ordene os resultados por **ano** crescente e **pontos** decrescente.**

```python
res_ano_asc = resultados_df['ano'].sort_values().index

resultados_df.loc[res_ano_asc]
```

```python
res_ano_desc = resultados_df['ano'].sort_values(ascending=False).index

resultados_df.loc[res_ano_desc]
```

**Crie uma coluna **podio** em **resultados**, indicando **Sim** quando a posição final for 1, 2 ou 3.**

```python
resultados_df['posicao_final'].value_counts()
```

```python
resultados_df['podio'] = 'Nao'

dnf = resultados_df['posicao_final'] == 'DNF'
resultados_df.loc[dnf, ['posicao_final']] = '99'

resultados_df[dnf]
```

```python
resultados_df['posicao_final'] = resultados_df['posicao_final'].astype(int)

podio = resultados_df['posicao_final'] <= 3
resultados_df[podio].head()
```

```python
resultados_df.loc[podio, 'podio'] = 'Sim'

resultados_df[podio].head()
```

**Renomeie colunas financeiras para nomes mais fáceis: **investimento_desenvolvimento_m USD** → **investimento_musd**; **verba_patrocinadores_m USD** → **patrocinio_musd**.**

```python
new_columns = {}
for col in financeiro_df.columns.values:
    new_columns[col] = col

new_columns
```

```python
new_columns['investimento_desenvolvimento_m USD'] = 'investimento_musd'
new_columns['verba_patrocinadores_m USD'] = 'patrocinio_musd'

new_columns
```

```python
financeiro_df.rename(columns=new_columns, inplace=True)
financeiro_df.head()
```
