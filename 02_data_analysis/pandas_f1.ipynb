#### **Explique por que **merge** com patrocinadores pode multiplicar linhas.**

```python
print(f'Número de elementos em Patrocinadores: {len(patrocinadores_df)}')
print('\nColunas de Patrocinadores:', patrocinadores_df.columns)

patrocinadores_df.head()
```

```python
teste_merge = pd.merge(financeiro_df, patrocinadores_df, on='ano', how='outer')

print('Número de elementos no Teste Merge entre Financeiro e Patrocinadores:', len(teste_merge))
teste_merge.head()
```

O número de elementos pode explodir porque o dataset de patrocinadores possui diversas linhas com mesmo ano e mesma equipe, já que uma mesma equipe consegue mais de um patrocinador em uma mesma temporada. Assim, várias informações provindas de Financeiro se tornam redundantes já que são anuais.

#### **Agregue patrocinadores por **ano** e **equipe** antes do merge. Depois junte com resultados.**

```python
financ_patrocinadores_df = pd.merge(financeiro_df, patrocinadores_df, on=['ano', 'equipe'], how='inner')

print('Número de elementos em Financeiro_Patrocinadores:', len(financ_patrocinadores_df))
financ_patrocinadores_df.head(7)
```
