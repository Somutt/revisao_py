### 3.3 Tratamento de nulos

Trate os nulos conforme a regra de negócio:

- **cidade**: preencher com **"Desconhecido"**
- **segmento**: preencher com **"Não informado"**
- **idade**: preencher com mediana
- **valor**: preencher com média

Depois, confira novamente com **isnull().sum()**.


```python
clientes_df.isnull().sum()
```

```python
cidade_nan = clientes_df['cidade'].isna()

print(f'Número de elementos nulos em cidade nos clientes: {len(clientes_df[cidade_nan])}')
clientes_df[cidade_nan].head()
```

```python
clientes_df.loc[cidade_nan, 'cidade'] = 'Desconhecido'

clientes_df[cidade_nan].head()
```

```python
segmento_nan = clientes_df['segmento'].isna()

print(f'Número de elementos nulos em segmento do cliente: {len(clientes_df[segmento_nan])}')
clientes_df[segmento_nan].head()
```

```python
clientes_df.loc[segmento_nan, 'segmento'] = 'Não informado'

clientes_df[segmento_nan].head()
```

```python
idades_nan = clientes_df['idade'].isna()

print(f'Mediana da idade dos clientes: {clientes_df['idade'].median():.2f}; Número de valores nulos: {len(clientes_df[idades_nan])}')
clientes_df[idades_nan].head()
```

```python
clientes_df.loc[idades_nan, 'idade'] = clientes_df['idade'].median()

clientes_df[idades_nan].head()
```

```python
clientes_df.isnull().sum()
```

```python
valor_nan = pedidos_df['valor'].isna()

print(f'Média dos valores dos pedidos: {pedidos_df['valor'].mean().round(2)}; Número de nulos: {len(pedidos_df[valor_nan])}')
pedidos_df[valor_nan].head()
```

```python
pedidos_df.loc[valor_nan, 'valor'] = pedidos_df['valor'].mean().round(2)

pedidos_df[valor_nan].head()
```

```python
lote2_valor_nan = lote_2_df['valor'].isna()
lote_2_df.loc[lote2_valor_nan, 'valor'] = lote_2_df['valor'].mean().round(2)

lote_2_df[lote2_valor_nan].head()
```
