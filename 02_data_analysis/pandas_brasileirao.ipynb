## 2 - Top Artilheiros

1. Descubra os 15 jogadores com mais gols.

2. Mostre:

3. Responda:
- jogadores mais caros necessariamente fazem mais gols?

```python
jogadores_df.head()
```

```python
clubes_df.head()
```

```python
# união dos datasets clube e jogadores apenas com as informações necessárias (nome, clube, gols e salario)
print(f'Número de entradas em jogadores_df antes do merge: {len(jogadores_df)}; entradas em clubes_df antes do merge: {len(clubes_df)}')

jogadores_clubes_df = pd.merge(jogadores_df, clubes_df, on='id_clube', how='left')
print(f'\nNúmero de entradas em jogadores_clubes_df: {len(jogadores_clubes_df)}')
jogadores_clubes_df.head()
```

```python
# isolando apenas as informações necessarias
relatorio_artilheiros = jogadores_clubes_df[['nome', 'clube', 'gols', 'salario']]
relatorio_artilheiros.head()
```

```python
# modificando o valor de salario para milhares e arredondando para 2 p/ facilitar leitura
relatorio_artilheiros['salario'] = relatorio_artilheiros['salario'] / 1000

relatorio_artilheiros.head()
```

```python
# arredondando salario p/ 2 e mudando nome da coluna
relatorio_artilheiros['salario'] = relatorio_artilheiros['salario'].round(2)
relatorio_artilheiros.rename(columns={'salario': 'salario_milhares'}, inplace=True)

relatorio_artilheiros.head()
```

```python
# agregando e ranqueando top 15 por gols
relatorio_agg = relatorio_artilheiros.groupby('nome').agg(
                        total_gols=('gols', 'sum'),
                        total_salario_milhares=('salario_milhares', 'sum')
                    )

relatorio_agg.sort_values('total_gols', ascending=False).head(15)
```

```python
relatorio_agg.sort_values('total_gols').head(15)
```

Como é possível notar, não só jogadores com salários considerados médios estão entre aqueles que mais marcaram gols (como o 1373 e 197), ambos com mais de 10 gols como também existem jogadores que não marcaram gols e que recebem salarios astronômicos (como o 1464 e 2699) de modo que o gasto salarial não aparece como fator decisivo daqueles jogadores que mais marcam gols.
