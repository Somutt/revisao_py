### 1.1. Operadores Aritiméticos

Um **operador aritmético** é um símbolo que representa um cálculo aritmético. Por exemplo, o sinal de mais, `+`, realiza uma adição.

````python
10 + 3
````

O sinal de menos, `-`, é o operador utilizado para designar a subtração.

````python
10 - 3
````

O sinal de esterisco, `*`, designa a multiplicação.

````python
10 * 3
````

A barra, `/`, é utilizado para designar a divisão.

````python
10 / 2
````

Para obter o resto da divisão, utiliza-se o operador `%`.

````python
10 % 2
````

Observe que o resultado da divisão é `5.0` em vez de `5`. Isso ocorre porque existem dois tipos de números em Python:

- **inteiros**, que representam números sem parte fracionária ou decimal, e
- **números de ponto flutuante**, que representam números inteiros e números com ponto decimal.

Se você adicionar, subtrair ou multiplicar dois números inteiros, o resultado será um número inteiro. Mas se você dividir dois números inteiros, o resultado será um número de ponto flutuante. Python fornece outro operador, `//`, para realizar a divisão de números inteiros. O resultado da divisão de números inteiros é sempre um número inteiro.

````python
10 // 2
````

Por fim, o operador `**` realiza a exponenciação; ou seja, eleva um número a uma potência.

````python
10 ** 2
````

Resumo dos operadores aritiméticos:
| Símbolo | Nome            |
|:-------:|-----------------|
| `+`     | Soma            |
| `-`     | Subtração       | 
| `*`     | Multiplicação   |
| `/`     | Divisão         |
| `%`     | Resto da Divisão|
| `**`    | Exponenciação   | 
| `//`    | Divisão Inteira |

Existem outros operadores aritiméticos: `<< `, `>>`, `&`, `|`, `~` e `^`. Todos esses operadores têm algo em comum: são operadores "bit a bit". Ou seja, operam com números (normalmente), mas em vez de tratar esse número como se fosse um único valor, tratam-no como se fosse uma sequência de bits, escrita em binário em complemento de dois. Porém, não trataremos desses operadores no escopo desse curso. Você pode obter mais inforações na documentação oficial do [Python](https://wiki.python.org/moin/BitwiseOperators).

### 1.2. Expressões

Uma coleção de operadores e números é chamada de expressão. Uma expressão pode conter qualquer número de operadores e números. Por exemplo, aqui está uma expressão que contém dois operadores.

````python
3 + 3 ** 2
````

Observe que a exponenciação ocorre antes da adição. Python segue a ordem de operações que você provavelmente aprendeu em uma aula de matemática: a exponenciação ocorre antes da multiplicação e da divisão, que ocorrem antes da adição e da subtração.

No exemplo a seguir, a multiplicação ocorre antes da adição.

````python
12 + 5 * 6
````

Se você quiser que a adição aconteça primeiro, você pode usar parênteses.

````python
(12 + 5) * 6
````

**Cada expressão** tem um **valor** associado. Por exemplo, a expressão `6 * 7` tem o valor `42`.

### 1.3. Funções Aritiméticas

Além dos operadores aritméticos, Python oferece algumas funções que trabalham com números. Por exemplo, a função `round` pega um número de ponto flutuante e o arredonda para o inteiro mais próximo.

````python
round(10.4)
````

````python
round(10.6)
````

A função `abs` calcula o valor absoluto de um número. Para um número positivo, o valor absoluto é o próprio número.

````python
abs(10.4)
````

Para um número negativo, o valor absoluto é positivo.

````python
abs(-10.4)
````

Quando usamos uma função como esta, dizemos que estamos **chamando** a função. Uma expressão que chama uma função é uma **chamada de função**.

Quando você chama uma função, os parênteses são obrigatórios. Se você os omitir, receberá uma mensagem de erro.

````python
abs 10.4
````

Você pode ignorar a primeira linha desta mensagem; ela não contém nenhuma informação que precisamos entender agora. A segunda linha é o código que contém o erro, com um acento circunflexo (`^`) abaixo para indicar onde o erro foi descoberto.

A última linha indica que se trata de um erro de sintaxe, o que significa que há algo errado com a estrutura da expressão. Neste exemplo, o problema é que uma chamada de função requer parênteses.

Vamos ver o que acontece se você omitir os parênteses e o valor.

````python
abs
````

Um nome de função por si só é uma expressão legal que possui um valor. Quando exibido, o valor indica que abs é uma função e inclui algumas informações adicionais.

### 1.4. Strings

Além de números, Python também pode representar sequências de letras, que são chamadas de strings porque as letras estão encadeadas como contas em um colar. Para escrever uma string, podemos colocar uma sequência de letras entre aspas retas.

````python
'Ola'
````

Também é permitido usar aspas duplas.

````python
"Ola"
````

Aspas duplas facilitam a escrita de uma string que contém um apóstrofo, que é o mesmo símbolo de uma aspa reta.

````python
"olho d'água"
````

Strings também podem conter espaços, pontuação e dígitos.

````python
"Bem, "
````

O operador `+` funciona com strings; ele une (concatena) duas strings em uma única string, o que é chamado de concatenação.

````python
"Bem, " + "esse é um mundo " + "pequeno!"
````

O operador `*` também funciona com strings; ele faz várias cópias de uma string e as concatena.

````python
"ola, " * 4
````

Os outros operadores aritméticos não funcionam com strings.

O Python fornece uma função chamada `len` que calcula o comprimento de uma string.

````python
len("Ola mundo!")
````

Observe que len conta as letras entre aspas, mas não as aspas.

Ao criar uma string, certifique-se de usar aspas retas. As aspas posteriores, também conhecidas como crase, causam um erro de sintaxe.

````python
`Ola`
````

### 1.5. Valores e Tipos

Até agora, vimos três tipos de valores:
- `2` é um inteiro,
- `42.0` é um número de ponto flutuante e
- `'Ola'` é uma string.

Um tipo de valor é chamado de **tipo**. Todo valor tem um tipo – ou às vezes dizemos que ele **pertence a** um tipo.

O Python fornece uma função chamada `type` que informa o tipo de qualquer valor. O tipo de um inteiro é `int`.

````python
type(10)
````

O tipo de um número de ponto flutuante é `float`.

````python
type(10.1)
````

E o tipo de uma string é `str`.

````python
type('Ola, Mundo!')
````

Os tipos `int`, `float` e `str` podem ser usados ​​como funções. Por exemplo, `int` pode receber um número de ponto flutuante e convertê-lo em um inteiro (sempre arredondando para baixo).

````python
int(10.3)
````

E `float` pode converter um inteiro em um valor de ponto flutuante.

````python
float(10)
````

Agora, aqui vai algo que pode ser confuso: o que acontece se você colocar uma sequência de dígitos entre aspas?

````python
'123'
````

Parece um número, mas na verdade é uma string.

````python
type('123')
````

Se você tentar usá-lo como um número, poderá obter um erro.

````python
'123' / 2
````

Este exemplo gera um `TypeError`, o que significa que os valores na expressão, chamados operandos, têm o tipo errado. A mensagem de erro indica que o operador `/` não suporta os tipos desses valores, que são `str` e `int`.

Se você tiver uma string que contém dígitos, você pode usar `int` para convertê-la em um inteiro.

````python
int('123') / 2
````

Se você tiver uma string que contém dígitos e um ponto decimal, você pode usar `float` para convertê-la em um número de ponto flutuante.

````python
float('10.2')
````

Ao escrever um número inteiro grande, você pode se sentir tentado a usar vírgulas entre grupos de dígitos, como em `1,000,000`. Esta é uma expressão válida em Python, mas o resultado não é um número inteiro.

````python
1,000,000
````

O Python interpreta `1,000,000` como uma sequência de números inteiros separados por vírgulas. Falaremos mais sobre esse tipo de sequência posteriormente.

Você pode usar sublinhados para facilitar a leitura de números grandes.

````python
1_000_000
````

### 1.6. Linguagens Natural e Formal

**Linguagens naturais** são as línguas que as pessoas falam, como português, inglês, espanhol e francês. Elas não foram projetadas por pessoas; elas evoluíram naturalmente.

**Linguagens formais** são linguagens projetadas por pessoas para aplicações específicas. Por exemplo, a notação que os matemáticos usam é uma linguagem formal que é particularmente boa para denotar relações entre números e símbolos. Da mesma forma, linguagens de programação são linguagens formais que foram projetadas para expressar cálculos.

Embora as linguagens formais e naturais tenham algumas características em comum, existem diferenças importantes:

- Ambiguidade: As linguagens naturais são cheias de ambiguidade, com a qual as pessoas lidam usando pistas contextuais e outras informações. As linguagens formais são projetadas para serem quase ou completamente inequívocas, o que significa que qualquer programa tem exatamente um significado, independentemente do contexto.

- Redundância: Para compensar a ambiguidade e reduzir mal-entendidos, as linguagens naturais usam redundância. Como resultado, elas costumam ser prolixas. As linguagens formais são menos redundantes e mais concisas.

- Literalidade: As linguagens naturais são cheias de expressões idiomáticas e metáforas. Linguagens formais significam exatamente o que dizem.

Como todos crescemos falando linguagens naturais, às vezes é difícil nos adaptarmos às linguagens formais. As linguagens formais são mais densas do que as linguagens naturais, por isso levam mais tempo para serem lidas. Além disso, a estrutura é importante, por isso nem sempre é melhor ler de cima para baixo, da esquerda para a direita. Por fim, os detalhes importam. Pequenos erros de ortografia e pontuação, que você pode evitar em linguagens naturais, podem fazer uma grande diferença em uma linguagem formal.

### 2.1. Variáveis

Uma **variável** é um nome que se refere a um valor. Para criar uma variável, podemos escrever uma instrução de atribuição como esta.

````python
n = 10
````

````python
mensagem = 'E agora algo completamente diferente'
````

Ao executar uma instrução de atribuição, não há saída. O Python cria a variável e atribui um valor a ela, mas a instrução de atribuição não tem efeito visível. No entanto, após criar uma variável, você pode usá-la como uma expressão. Assim, podemos exibir o valor de `mensagem` assim:

````python
mensagem
````

Você também pode usar uma variável como parte de uma expressão com operadores aritméticos.


````python
n + 25
````

````python
2 * pi
````

E você pode usar uma variável quando chamar uma função.

````python
round(pi)
````

### 2.3. Nome de Variáveis

Os nomes de variáveis ​​podem ser tão longos quanto você quiser. Eles podem conter letras e números, mas não podem começar com um número. É permitido usar letras maiúsculas, mas é convencional usar apenas letras minúsculas para nomes de variáveis.

A única pontuação que pode aparecer em um nome de variável é o caractere de sublinhado, `_`. Ele é frequentemente usado em nomes com várias palavras, como `seu_nome` ou `velocidade_do_ar_de_andorinha_sem_carga`.

### 2.5. Expressões e Comandos

Até agora, vimos alguns tipos de expressões. Uma expressão pode ser um único valor, como um inteiro, um número de ponto flutuante ou uma string. Também pode ser uma coleção de valores e operadores. ### 2.7. Argumentos

Quando você chama uma função, a expressão entre parênteses é chamada de `argumento`. Algumas das funções que vimos até agora aceitam apenas um argumento, como `int`.

````python
int('101')
````

Alguns usam dois, como o `math.pow`.

````python
math.pow(2, 4)
````

Algumas podem receber argumentos adicionais opcionais. Por exemplo, `int` pode receber um segundo argumento que especifica a base do número.

````python
int('101', 2)
````

A sequência de dígitos `101` na base 2 representa o número 5 na base 10.

`round` também aceita um segundo argumento opcional, que é o número de casas decimais para arredondar.

````python
round(math.pi, 4)
````

Algumas funções podem receber qualquer número de argumentos, como `print`.

````python
print('Qualquer', 'número', 'de', 'argumentos')
````

````python
float('101.0', 2)
````

````python
math.pow(2)
````

E se você fornecer um argumento com um tipo que a função não pode manipular, isso também é um `TypeError`.

````python
math.sqrt('81')
````

Esse tipo de verificação pode ser irritante quando você está começando, mas ajuda a detectar e corrigir erros.

### 2.8. Comentários

À medida que os programas se tornam maiores e mais complexos, tornam-se mais difíceis de ler. Linguagens formais são densas e, muitas vezes, é difícil analisar um trecho de código e descobrir o que ele está fazendo e por quê.

Por esse motivo, é uma boa ideia adicionar notas aos seus programas para explicar em linguagem natural o que o programa está fazendo. Essas notas são chamadas de **comentários** e começam com o símbolo `#`.

````python
# número de segundos em 42:42
segundos = 42 * 60 + 42
````

Neste caso, o comentário aparece sozinho em uma linha. Você também pode colocar comentários no final de uma linha:

````python
milhas = 10 / 1.61 # 10 quilômetros em milhas
````

Tudo, desde o `#` até o final da linha, é ignorado — não afeta a execução do programa.

Comentários são mais úteis quando documentam características não óbvias do código. É razoável supor que o leitor consiga descobrir *o que* o código faz; é mais útil explicar *o porquê*.

Este comentário é redundante com o código e inútil:

````python
v = 8 # atribui 8 a v
````

Este comentário contém informações úteis que não estão no código:

````python
v = 8 # velocidade em milhas por hora
````

Bons nomes de variáveis ​​podem reduzir a necessidade de comentários, mas nomes longos podem tornar expressões complexas difíceis de ler, então há uma compensação.

### 3.5. Repetição (`for` e `while`)

Se quisermos exibir mais de um verso, podemos usar a instrução `for`. 

```python
for variavel in minha_string_ou_lista :
  # faça alguma coisa com variavel
```

A `variavel` que aparece logo após a palavra `for` vai receber o valor de cada elemento da `minha_string_ou_lista` por vez.

Aqui está um exemplo simples.

````python
for i in range(2):
    print(i)
````

A primeira linha é um cabeçalho que termina com dois pontos. A segunda linha é o corpo, que precisa ser recuado.

O cabeçalho começa com a palavra-chave `for`, uma nova variável chamada `i` e outra palavra-chave, `in`. Ele usa a função `range` para criar uma sequência de dois valores, que são `0` e `1`. Em Python, quando começamos a contar, geralmente começamos do `0`.

Quando o comando `for` é executado, ele atribui o primeiro valor de `range` a `i` e, em seguida, executa a função `print` no corpo, que exibe `0`.

Quando chega ao final do corpo, ela retorna ao cabeçalho, e é por isso que esse comando é chamado de **laço (loop)**. Na segunda vez que passa pelo loop, ela atribui o próximo valor de `range` a `i` e o exibe. Então, como esse é o último valor de `range`, o loop termina.

Veja como podemos usar um loop `for` para imprimir dois versos da música.

````python
for i in range(2):
    print("Verso", i)
    print_verso()
    print()
````

Você pode colocar um loop for dentro de uma função. Por exemplo, `print_n_versos` recebe um parâmetro chamado `n`, que deve ser um inteiro, e exibe o número de versos fornecido.

````python
def print_n_versos(n):
    for i in range(n):
        print_verso()
        print()

print_n_versos(4)
````

Neste exemplo, não usamos `i` no corpo do loop, mas de qualquer forma, tem que haver um nome de variável no cabeçalho.

O comando `while` (que significa enquanto) é outro tipo de estrutura de repetição em Python.

Ele é muito útil quando precisamos repetir um trecho de código.

O `while` é parecido com um comando `if`: ele vai executar o código dentro dele se determinada condição for verdadeira. A diferença é que o comando `while` continuará executando **enquanto** aquela condição for verdadeira.

Normalmente ele tem a seguinte sintaxe:
```python
while condicao :
  comandos
```  

Estruturas de repetição são também chamadas de **laço** ou **loop**.

Exemplo de uso do comando `while`:

````python
cont = 0
while cont < 5:
  print("Olá, eu sou um laço while e cont é", cont)
  cont = cont + 1
````

No exemplo acima, a linha 2 decide quando o loop será executado. Portanto, *enquanto cont for menor do 5*, o loop vai continuar a executar. A linha 4 aumenta cont em 1, isso acontece repetidamente até que cont seja igual a 5.

### 3.8. Docstrings

Uma **docstring** é uma string no início de uma função que explica a interface ("doc" é a abreviação de "documentação"). Aqui está um exemplo:

````python
def cat_twice(part1, part2):
    """Concatena os dois argumentos (as partes) e imprime duas vezes o resultado da concatenação.
    
    part1: é uma string da primeira parte
    part2: é uma string da segunda parte
    """
    cat = part1 + part2
    print_twice(cat)
````

Por convenção, docstrings são strings entre aspas triplas, também conhecidas como **strings multilinhas**, pois as aspas triplas permitem que a string ocupe mais de uma linha.

Uma docstring deve:

- Explicar concisamente o que a função faz, sem entrar em detalhes sobre seu funcionamento,

- Explicar o efeito de cada parâmetro no comportamento da função e

- Indicar o tipo de cada parâmetro, caso não seja óbvio.

Escrever esse tipo de documentação é uma parte importante do design de interfaces de funções. Uma interface bem projetada deve ser simples de explicar; se você tiver dificuldade em explicar uma de suas funções, talvez a interface possa ser melhorada.

### 3.9. Por que Funções?

Talvez ainda não esteja claro por que vale a pena dividir um programa em funções. Há vários motivos:

- Criar uma nova função oferece a oportunidade de nomear um grupo de instruções, o que torna seu programa mais fácil de ler e depurar.

- As funções podem tornar um programa menor, eliminando código repetitivo. Posteriormente, se você fizer uma alteração, precisará fazê-la apenas em um lugar.

- Dividir um programa longo em funções permite depurar as partes uma de cada vez e, em seguida, reuni-las em um todo funcional.

- Funções bem projetadas costumam ser úteis para muitos programas. Depois de escrever e depurar uma, você pode reutilizá-la.

### 4.4. O Comando if

Para escrever programas úteis, quase sempre precisamos da capacidade de verificar condições e alterar o comportamento do programa de acordo. **Instruções condicionais** nos dão essa capacidade. A forma mais simples é a instrução `if`:

````python
if x > 0:
    print('x é positivo')
````

`if` é uma palavra-chave do Python. Instruções `if` têm a mesma estrutura que definições de função: um cabeçalho seguido por uma instrução ou sequência de instruções recuada, chamada de `bloco`.

A expressão booleana após `if` é chamada de **condição**. Se for verdadeira, as instruções no bloco recuado são executadas. Caso contrário, não são executadas.

Não há limite para o número de instruções que podem aparecer no bloco, mas deve haver pelo menos uma. ### 4.5. A cláusula else

Uma instrução `if` pode ter uma segunda parte, chamada cláusula `else`. A sintaxe é semelhante a esta:

````python
if x % 2 == 0:
    print('x é par')
else:
    print('x é ímpar')
````

Se a condição for verdadeira, a primeira instrução recuada é executada; caso contrário, a segunda instrução recuada é executada.

Neste exemplo, se `x` for par, o resto da divisão de `x` por `2` será `0`, então a condição é verdadeira e o programa exibe que `x é par`. Se `x` for ímpar, o resto será `1`, então a condição é falsa e o programa exibe que `x é ímpar`.

Como a condição deve ser verdadeira ou falsa, exatamente uma das alternativas será executada. As alternativas são chamadas de **desvios** (desvios condicionais ou *braches*).

### 4.6. Condicionais Encadeados

Às vezes, há mais de duas possibilidades e precisamos de mais de dois desvios/ramos. Uma maneira de expressar um cálculo como esse é uma condicional encadeada, que inclui uma cláusula `elif`.

````python
if x < y:
    print('x é menor que y')
elif x > y:
    print('x é maior que y')
else:
    print('x e y são iguais')
````

`elif` é uma abreviação de "else if". Não há limite para o número de cláusulas `elif`. Se houver uma cláusula `else`, ela precisa estar no final, mas não precisa haver uma.

Cada condição é verificada em ordem. Se a primeira for falsa, a próxima é verificada e assim por diante. Se uma delas for verdadeira, o desvio correspondente é executado e a instrução `if` termina. Mesmo que mais de uma condição seja verdadeira, apenas o primeiro desvio verdadeiro é executado.

### 4.7. Condicionais Aninhados

Uma condicional também pode ser aninhada dentro de outra. Poderíamos ter escrito o exemplo da seção anterior assim:

````python
if x == y:
    print('x e y são iguais')
else:
    if x < y:
        print('x é menor que y')
    else:
        print('x e y são iguais')
````

A instrução `if` externa contém dois desvios/ramos. O primeiro desvio/ramo contém uma instrução simples. O segundo desvio/ramo contém outra instrução `if`, que possui dois desvios/ramos próprios. Esses dois desvios/ramos são instruções simples, embora também pudessem ser instruções condicionais.

Embora a indentação das instruções torne a estrutura aparente, **condicionais aninhadas** podem ser difíceis de ler.

Operadores lógicos geralmente fornecem uma maneira de simplificar instruções condicionais aninhadas. Aqui está um exemplo com uma condicional aninhada.

````python
if 0 < x:
    if x < 10:
        print('x é um número positivo de um único dígito.')
````

A instrução `print` só será executada se passarmos por ambas as condicionais, então obtemos o mesmo efeito com o operador `and`.

````python
if 0 < x and x < 10:
    print('x é um número positivo de um único dígito.')
````

Para esse tipo de condição, o Python fornece uma opção mais concisa:

````python
if 0 < x < 10:
    print('x é um número positivo de um único dígito.')
````

### 4.8. Recursão

É legal que uma função chame a si mesma. Pode não ser óbvio por que isso é bom, mas acaba sendo uma das coisas mais mágicas que um programa pode fazer. Aqui está um exemplo.

````python
def countdown(n):
    if n <= 0:
        print('Decolagem!')
    else:
        print(n)
        countdown(n-1)
````

Se `n` for `0` ou negativo, `countdown` emite a palavra "Decolagem!". Caso contrário, ela imprime `n` e então chama a si mesmo, passando `n-1` como argumento.

Veja o que acontece quando chamamos esta função com o argumento `3`.

````python
countdown(3)
````

A execução de `countdown` começa com `n=3` e, como `n` é maior
que `0`, exibe `3` e, em seguida, chama a si mesmo\...

> A execução de `countdown` começa com `n=2` e, como `n` é
> maior que `0`, exibe `2` e, em seguida, chama a si mesmo\...
>
> > A execução de `countdown` começa com `n=1` e, como `n` é
> > maior que `0`, exibe `1` e, em seguida, chama a si mesmo\...
> >
> > > A execução de `countdown` começa com `n=0` e, como `n` não é
> > > maior que `0`, exibe "Decolagem!" e retorna.
> >
> > A `countdown` que obteve `n=1` retorna.
>
> A `countdown` que obteve `n=2` retorna.

A `countdown` que obteve `n=3` retorna.

Uma função que chama a si mesma é **recursiva**. Como outro exemplo, podemos escrever uma função que imprime uma string `n` vezes.

````python
def print_n_vezes(string, n):
    if n > 0:
        print(string)
        print_n_vezes(string, n-1)
````

Se `n` for positivo, `print_n_vezes` exibe o valor de `string` e, em seguida, chama a si mesma, passando `string` e `n-1` como argumentos.

Se `n` for `0` ou negativo, a condição é falsa e `print_n_vezes` não faz nada.

Veja como funciona.

````python
print_n_vezes('Spam', 4)
````

Para exemplos simples como este, provavelmente é mais fácil usar um laço `for`. Mas veremos exemplos mais adiante que são difíceis de escrever com um laço `for` e fáceis de escrever com recursão, então é bom começar cedo.

### 4.9. Recursão com Valores de Retorno

Agora que podemos escrever funções com valores de retorno, podemos escrever funções recursivas com valores de retorno e, com essa capacidade, ultrapassamos um limite importante: o subconjunto do Python que temos agora é Turing completo, o que significa que podemos executar qualquer computação que possa ser descrita por um algoritmo.

Para demonstrar recursão com valores de retorno, avaliaremos algumas funções matemáticas definidas recursivamente. Uma definição recursiva é semelhante a uma definição circular, no sentido de que a definição se refere ao objeto que está sendo definido.

A definição da função fatorial, denotada pelo símbolo $!$, pode ser definida como segue:

$$\begin{aligned}
0! &= 1 \\
n! &= n~(n-1)!
\end{aligned}$$ 

Esta definição diz que o fatorial de $0$ é $1$, e o fatorial de qualquer outro valor, $n$, é $n$ multiplicado pelo fatorial de $n-1$.

Se você consegue escrever uma definição recursiva de algo, você consegue escrever um programa em Python para avaliá-la. Seguindo um processo de desenvolvimento incremental, começaremos com uma função que recebe `n` como parâmetro e sempre retorna `0`.

````python
def fatorial(n):
    return 0
````

Agora vamos adicionar a primeira parte da definição – se o argumento for `0`, tudo o que precisamos fazer é retornar `1`:

````python
def fatorial(n):
    if n == 0:
        return 1
    else:
        return 0
````

Agora vamos preencher a segunda parte – se `n` não for `0`, temos que fazer uma chamada recursiva para encontrar o fatorial de `n-1` e então multiplicar o resultado por `n`:

````python
def fatorial(n):
    if n == 0:
        return 1
    else:
        return n * fatorial(n-1)
````

Se chamarmos `fatorial` com o valor `3`:

Como `3` não é `0`, pegamos o segundo ramo e calculamos o fatorial
de `n-1`\...

> Como `2` não é `0`, pegamos o segundo ramo e calculamos o
> fatorial de `n-1`\...
>
> > Como `1` não é `0`, pegamos o segundo ramo e calculamos o
> > fatorial de `n-1`\...
> >
> > > Como `0` é igual a `0`, pegamos o primeiro ramo e retornamos `1` sem
> > > fazer mais chamadas recursivas.
> >
> > O valor de retorno, `1`, é multiplicado por `n`, que é `1`, e o
> > resultado é retornado.
>
> O valor de retorno, `1`, é multiplicado por `n`, que é `2`, e o resultado
> é retornado.

O valor de retorno `2` é multiplicado por `n`, que é `3`, e o resultado,
`6`, torna-se o valor de retorno da chamada de função que iniciou todo o
processo.

### 4.10. Recursão Infinita

Se uma recursão nunca atingir um caso base, ela continuará fazendo chamadas recursivas para sempre, e o programa nunca terminará. Isso é conhecido como **recursão infinita** e geralmente não é uma boa ideia. Aqui está uma função mínima com recursão infinita.

````python
def recursivo():
    recursivo()
````

Cada vez que `recursivo` é chamado, ele chama a si mesmo, o que cria outro quadro. Em Python, há um limite para o número de *frames* que podem estar na *pilha* ao mesmo tempo. Se um programa exceder o limite, ocorrerá um erro de execução.

````python
recursivo()
````

O traceback indica que havia quase 3000 frames na pilha quando o erro ocorreu.

Se você encontrar uma recursão infinita por acidente, revise sua função para confirmar se há um caso base que não faça uma chamada recursiva. E se houver um caso base, verifique se há garantia de que você o alcançará.

Implemente uma versão recursiva de `fibonacci` em Python.

### 5.8. Expressões Regulares

Se soubermos exatamente qual sequência de caracteres estamos procurando, podemos usar o operador `in` para encontrá-la e o método `replace` para substituí-la. Mas existe outra ferramenta, chamada expressão regular, que também pode realizar essas operações – e muito mais.

Para demonstrar, começarei com um exemplo simples e avançaremos. Suponha, novamente, que queremos encontrar todas as linhas que contêm uma palavra específica. Para variar, vamos procurar referências ao personagem-título do livro, o Conde Drácula. Aqui está uma linha que o menciona.

````python
texto = "I am Dracula; and I bid you welcome, Mr. Harker, to my house."
````

E aqui está o padrão que usaremos para pesquisar.

````python
pattern = 'Dracula'
````

Um módulo chamado `re` fornece funções relacionadas a expressões regulares. Podemos importá-lo assim e usar a função de `search` para verificar se o padrão aparece no texto.

````python
import re

result = re.search(pattern, texto)
result
````

Se o padrão aparecer no texto, `search` retornará um objeto `Match` que contém os resultados da pesquisa. Entre outras informações, ele possui uma variável chamada `string` que contém o texto pesquisado.

````python
result.string
````

Ele também fornece um método chamado `group` que retorna a parte do texto que corresponde ao padrão.

````python
result.group()
````

E fornece um método chamado `span` que retorna o índice no texto onde o padrão começa e termina.

````python
result.span()
````

Se o padrão não aparecer no texto, o valor de retorno de `search` será `None`.

````python
result = re.search('Count', texto)
print(result)
````

Juntando tudo isso, aqui está uma função que percorre as linhas do livro até encontrar uma que corresponda ao padrão fornecido e retorna o objeto `Match`.

````python
def find_first(pattern):
    for line in open('./files/pg345_cleaned.txt'):
        result = re.search(pattern, line)
        if result != None:
            return result
````

Podemos usá-lo para encontrar a primeira menção de um personagem.

````python
result = find_first('Harker')
result.string
````

Neste exemplo, não precisamos usar expressões regulares – poderíamos ter feito a mesma coisa mais facilmente com o operador `in`. Mas expressões regulares podem fazer coisas que o operador `in` não pode.

Por exemplo, se o padrão incluir o caractere de barra vertical, `'|'`, ele pode corresponder à sequência à esquerda ou à sequência à direita. Suponha que queremos encontrar a primeira menção de Mina Murray no livro, mas não temos certeza se ela é mencionada pelo primeiro nome ou pelo sobrenome. Podemos usar o padrão a seguir, que corresponde a qualquer um dos nomes.

````python
pattern = 'Mina|Murray'
result = find_first(pattern)
result.string
````

Podemos usar um padrão como este para ver quantas vezes um personagem é mencionado por qualquer nome. Aqui está uma função que percorre o livro e conta o número de linhas que correspondem ao padrão fornecido.

````python
def count_matches(pattern):
    count = 0
    for line in open('./files/pg345_cleaned.txt'):
        result = re.search(pattern, line)
        if result != None:
            count += 1
    return count
````

Agora vamos ver quantas vezes Mina é mencionada.

````python
count_matches('Mina|Murray')
````

O caractere especial `'^'` corresponde ao início de uma string, então podemos encontrar uma linha que começa com um padrão determinado.

````python
result = find_first('^Dracula')
result.string
````

E o caractere especial `'$'` corresponde ao final de uma string, então podemos encontrar uma linha que termina com um padrão dado (ignorando a nova linha no final).

````python
result = find_first('Harker$')
result.string
````

Bram Stoker nasceu na Irlanda e, quando *Drácula* foi publicado em 1897, ele morava na Inglaterra. Portanto, esperaríamos que ele usasse a grafia britânica de palavras como "centre" e "colour".
````python
pattern = 'colou?r'
````

Este padrão combina "colour" com o `'u'` ou "color" sem ele.

````python
result = find_first(pattern)
line = result.string
line
````

Novamente, como esperado, ele usou a grafia britânica.

Agora, suponha que queremos produzir uma edição do livro com a grafia americana. Podemos usar a função `sub` no módulo `re`, que faz a **substituição de strings**.

````python
re.sub(pattern, 'color', line)
````

O primeiro argumento é o padrão que queremos encontrar e substituir, o segundo é o que queremos usar para substituí-lo e o terceiro é a string que queremos pesquisar. No resultado, você pode ver que "colour" foi substituído por "color".

## 6. Listas, Dicionários e Tuplas

### 6.1. Listas

#### 6.1.1. Uma Lista é uma Sequência

Assim como uma string, uma lista é uma sequência de valores. Em uma string, os valores são caracteres; em uma lista, eles podem ser de qualquer tipo. Os valores em uma lista são chamados de **elementos**.

Existem várias maneiras de criar uma nova lista; a mais simples é colocar os elementos entre colchetes (`[` e `]`). Por exemplo, aqui está uma lista de dois inteiros.

````python
numeros = [42, 123]
numeros
````

E aqui está uma lista de três strings.

````python
queijos = ['Cheddar', 'Edam', 'Gouda']
queijos
````

Os elementos de uma lista não precisam ser do mesmo tipo. A lista a seguir contém uma string, um float, um inteiro e até mesmo outra lista.

````python
t = ['spam', 2.0, 5, [10, 20]]
t
````

Uma lista dentro de outra lista é **aninhada**.

Uma lista que não contém elementos é chamada de lista vazia; você pode criar uma com colchetes vazios, `[]`.

````python
vazia = []
vazia
````

A função `len` retorna o comprimento de uma lista.

````python
len(queijos)
````

O comprimento de uma lista `vazia` é `0`.

#### 6.1.2. Listas são Mutáveis

Para ler um elemento de uma lista, podemos usar o operador de colchetes. O índice do primeiro elemento é `0`.

````python
queijos[0]
````

Ao contrário das strings, as listas são mutáveis. Quando o operador de colchetes aparece à esquerda de uma atribuição, ele identifica o elemento da lista que será atribuído.

````python
numeros[1] = 17
numeros
````

O segundo elemento de `numeros`, que costumava ser `123`, agora é `17`.

Índices de lista funcionam da mesma forma que índices de string:

- Qualquer expressão inteira pode ser usada como índice.

- Se você tentar ler ou escrever um elemento que não existe, receberá um `IndexError`.

- Se um índice tiver um valor negativo, ele conta para trás a partir do final da lista.

O operador `in` funciona em listas – ele verifica se um determinado elemento aparece em algum lugar da lista.

````python
'Edam' in queijos
````

````python
'Wensleydale' in queijos
````

Embora uma lista possa conter outra lista, a lista aninhada ainda conta como um único elemento – portanto, na lista a seguir, há apenas quatro elementos.

````python
t = ['spam', 2.0, 5, [10, 20]]
len(t)
````

E `10` não é considerado um elemento de `t` porque é um elemento de uma lista aninhada, não `t`.

### 6.2. Dicionários

#### 6.2.1. Dicionários como Mapeamentos

Um **dicionário** é como uma lista, mas mais geral. Em uma lista, os índices precisam ser inteiros; em um dicionário, eles podem ser de (quase) qualquer tipo. Por exemplo, suponha que façamos uma lista de palavras numéricas, como esta.

````python
lista = ['zero', 'um', 'dois']
````

Podemos usar um número inteiro como índice para obter a palavra correspondente.

````python
lista[1]
````

Mas suponha que queiramos ir na outra direção e procurar uma palavra para obter o inteiro correspondente. Não podemos fazer isso com uma lista, mas podemos com um dicionário. Começaremos criando um dicionário vazio e atribuindo-lhe números.

````python
numeros = {}
numeros
````

As chaves, `{}`, representam um dicionário vazio. Para adicionar itens ao dicionário, usaremos colchetes.

````python
numeros['zero'] = 0
````

Esta atribuição adiciona ao dicionário um **item**, que representa a associação de uma **chave** e um **valor**. Neste exemplo, a chave é a string `'zero'` e o valor é o inteiro `0`. Se exibirmos o dicionário, veremos que ele contém um item, que contém uma chave e um valor separados por dois pontos, `:`.

````python
numeros
````

Podemos adicionar mais itens como este.

````python
numeros['um'] = 1
numeros['dois'] = 2
numeros
````

Agora, o dicionário contém três itens.

Para consultar uma chave e obter o valor correspondente, usamos o operador de colchetes.

````python
numeros['dois']
````

Se a chave não estiver no dicionário, obteremos um `KeyError`.

````python
numeros['tres']
````

A função `len` funciona em dicionários; ela retorna o número de itens.

````python
len(numeros)
````

Em linguagem matemática, um dicionário representa um mapeamento de chaves para valores, então também podemos dizer que cada chave "mapeia" para um valor. Neste exemplo, cada palavra numérica mapeia para o inteiro correspondente.

#### 6.2.2. Criando Dicionários

Na seção anterior, criamos um dicionário vazio e adicionamos itens um de cada vez usando o operador de colchetes. Em vez disso, poderíamos ter criado o dicionário de uma só vez, assim.

````python
numeros = {'zero': 0, 'um': 1, 'dois': 2}
numeros
````

Cada item consiste em uma chave e um valor separados por dois pontos. Os itens são separados por vírgulas e colocados entre chaves.

Geralmente é útil fazer uma cópia antes de executar operações que modificam dicionários.

#### 6.2.3. Iterando sobre Dicionários

Se você usar um dicionário em uma instrução for, ele percorrerá as chaves do dicionário. Para demonstrar, vamos criar um dicionário que conte as letras de `'banana'`.

````python
def value_counts(string):
    counter = {}
    for letra in string:
        if letra not in counter:
            counter[letra] = 1
        else:
            counter[letra] += 1
    return counter
````

````python
counter = value_counts('banana')
counter
````

O loop a seguir imprime as chaves, que são as letras.

````python
for key in counter:
    print(key)
````

Para imprimir os valores, podemos usar o método `values`.

````python
for value in counter.values():
    print(value)
````

Para imprimir as chaves e os valores, podemos percorrer as chaves e procurar os valores correspondentes.

````python
for key in counter:
    value = counter[key]
    print(key, value)
````

### 6.3. Tuplas

#### 6.2.1. Tuplas são Parecidas com Listas

Uma tupla é uma sequência de valores. Os valores podem ser de qualquer tipo e são indexados por números inteiros, portanto, tuplas são muito parecidas com listas. A diferença importante é que tuplas são imutáveis.

Para criar uma tupla, você pode escrever uma lista de valores separados por vírgulas.

````python
t = 'l', 'u', 'p', 'i', 'n'
print(t)
type(t)
````

Embora não seja necessário, é comum colocar tuplas entre parênteses.

````python
t = ('l', 'u', 'p', 'i', 'n')
print(t)
type(t)
````

Para criar uma tupla com um único elemento, você precisa incluir uma vírgula final.

````python
t1 = 'p',
print(t1)
type(t1)
````

Um único valor entre parênteses não é uma tupla.

````python
t2 = ('p')
type(t2)
````

Outra maneira de criar uma tupla é com a função interna `tuple`. Sem argumentos, ela cria uma tupla vazia.

````python
t = tuple()
t
````

Se o argumento for uma sequência (string, lista ou tupla), o resultado será uma tupla com os elementos da sequência.

````python
t = tuple('lupin')
t
````

Como tupla é o nome de uma função interna, evite usá-la como nome de variável.

A maioria dos operadores de lista também funciona com tuplas. Por exemplo, o operador de colchetes indexa um elemento.

````python
t[0]
````

E o operador de fatia seleciona um intervalo de elementos.

````python
t[1:3]
````

O operador `+` concatena tuplas.

````python
tuple('lup') + ('i', 'n')
````

E o operador `*` duplica uma tupla um determinado número de vezes.

````python
tuple('spam') * 2 
````

A função `sorted` funciona com tuplas, mas o resultado é uma lista, não uma tupla.

````python
sorted(t)
````

A função `reversed` também funciona com tuplas.

````python
reversed(t)
````

O resultado é um objeto `reversed`, que podemos converter em uma lista ou tupla.

````python
tuple(reversed(t))
````

Com base nos exemplos até agora, pode parecer que tuplas são a mesma coisa que listas.

#### 6.2.2. Tuplas são Imutáveis

Se você tentar modificar uma tupla com o operador de colchetes, você receberá um `TypeError`.

````python
t[0] = 'L'
````

Você pode colocar uma tupla de variáveis ​​no lado esquerdo de uma atribuição e uma tupla de valores no lado direito.

Os valores são atribuídos às variáveis ​​da esquerda para a direita – neste exemplo, `a` obtém o valor `1` e `b` obtém o valor `2`. Podemos exibir os resultados assim:

````python
a, b
````

De forma mais geral, se o lado esquerdo de uma atribuição for uma tupla, o lado direito pode ser qualquer tipo de sequência – string, lista ou tupla. Por exemplo, para dividir um endereço de e-mail em um nome de usuário e um domínio, você poderia escrever:

````python
email = 'monty@python.org'
username, domain = email.split('@')
````

O valor de retorno de `split` é uma lista com dois elementos – o primeiro elemento é atribuído ao `username`, o segundo ao `domain`.

````python
username, domain
````

O número de variáveis ​​à esquerda e o número de valores à direita devem ser os mesmos – caso contrário, você obterá um `ValueError`.

````python
a, b = 1, 2, 3
````

A atribuição de tuplas é útil se você quiser trocar os valores de duas variáveis. Com atribuições convencionais, você precisa usar uma variável temporária, como esta:

````python
temp = a
a = b
b = temp
````

Isso funciona, mas com a atribuição de tupla podemos fazer a mesma coisa sem uma variável temporária.

````python
print(a, b)
a, b = b, a
print(a, b)
````

Isso funciona porque todas as expressões do lado direito são avaliadas antes de qualquer atribuição.

Também podemos usar a atribuição de tuplas em uma instrução `for`. Por exemplo, para percorrer os itens de um dicionário, podemos usar o método `items`.

````python
d = {'um': 1, 'dois': 2}

for item in d.items():
    key, value = item
    print(key, '->', value)
````

A cada repetição do loop, o `item` recebe uma tupla que contém uma chave e o valor correspondente.

Podemos escrever este loop de forma mais concisa, assim:

````python
for key, value in d.items():
    print(key, '->', value)
````

A cada vez que passa pelo loop, uma chave e o valor correspondente são atribuídos diretamente à `key` e ao `value`.

#### 6.2.5. Empacotamento de Argumentos

Funções podem receber um número variável de argumentos. Um nome de parâmetro que começa com o operador `*` **agrupa** os argumentos em uma tupla. Por exemplo, a função a seguir recebe qualquer número de argumentos e calcula sua média aritmética – ou seja, a soma deles dividida pelo número de argumentos.

````python
def mean(*args):
    return sum(args) / len(args)
````

O parâmetro pode ter o nome que você quiser, mas `args` é o nome convencional. Podemos chamar a função assim:

````python
mean(1, 2, 3)
````

Se você tiver uma sequência de valores e quiser passá-los para uma função como múltiplos argumentos, pode usar o operador `*` para **desempacotar** a tupla. Por exemplo, `divmod` recebe exatamente dois argumentos – se você passar uma tupla como parâmetro, receberá um erro.

````python
t = (7, 3)
divmod(t)
````

Mesmo que a tupla contenha dois elementos, ela conta como um único argumento. Mas se você descompactar a tupla, ela será tratada como dois argumentos.

````python
divmod(*t)
````

Empacotar e desempacotar pode ser útil se você quiser adaptar o comportamento de uma função existente. Por exemplo, esta função recebe qualquer número de argumentos, remove o menor e o maior e calcula a média dos demais.

````python
def trimmed_mean(*args):
    low, high = min_max(args)
    trimmed = list(args)
    trimmed.remove(low)
    trimmed.remove(high)
    return mean(*trimmed)
````

Primeiro, ele usa `min_max` para encontrar os elementos mais baixos e mais altos. Em seguida, converte `args` em uma lista para poder usar o método `remove`. Por fim, ele descompacta a lista para que os elementos sejam passados ​​para `mean` como argumentos separados, em vez de uma única lista.

Aqui está um exemplo que mostra o efeito.

````python
mean(1, 2, 3, 10)
````

````python
trimmed_mean(1, 2, 3, 10)
````

Esse tipo de média `'aparada'` é usado em alguns esportes com julgamento subjetivo – como mergulho e ginástica – para reduzir o efeito de um juiz cuja pontuação se desvia dos outros.

Uma cifra de César é uma forma fraca de criptografia que envolve deslocar cada letra por um número fixo de casas no alfabeto, retornando ao início, se necessário. Por exemplo, `'a'` deslocado em 2 casas é `'c'` e `'z'` deslocado em 1 casa é `'a'`.

````python
# escreva aqui sua resposta
letters = 'abcdefghijklmnopqrstuvwxyz'
numbers = range(len(letters))
letter_map = dict(zip(letters, numbers))

def shift_word(word, shift):
    encrypted_word = []
    for c in word:
        new_index = (letter_map[c] + shift) % len(letters)
        encrypted_word.append(letters[new_index])
    
    return ''.join(encrypted_word)

print(shift_word('cheer', 7))
print(shift_word('melon', 16))

texto = "ola mundo cruel"

c = []
for palavra in texto.split():
    c.append(shift_word(palavra, 3))

print(' '.join(c))
````

## 7. Módulos e Pacotes

### 7.1. Scripts como Módulo

Ao sair e entrar de novo no interpretador Python, as definições anteriores (funções e variáveis) são perdidas. Portanto, se quiser escrever um programa maior, será mais eficiente usar um editor de texto para preparar as entradas para o interpretador, e executá-lo usando o arquivo como entrada. Isso é conhecido como criar um *script*. Se o programa se torna ainda maior, é uma boa prática dividi-lo em arquivos menores, para facilitar a manutenção. Também é preferível usar um arquivo separado para uma função que você escreveria em vários programas diferentes, para não copiar a definição de função em cada um deles.

Para permitir isso, Python tem uma maneira de colocar as definições em um arquivo e então usá-las em um script ou em uma execução interativa do interpretador. Tal arquivo é chamado de *módulo*; definições de um módulo podem ser importadas para outros módulos, ou para o módulo *principal* (a coleção de variáveis a que você tem acesso num script executado como um programa).

Um módulo é um arquivo contendo definições e instruções Python. O nome do arquivo é o nome do módulo acrescido do sufixo `.py`. Dentro de um módulo, o nome do módulo (como uma string) está disponível como o valor da variável global `__name__`. Por exemplo, use seu editor de texto favorito para criar um arquivo chamado `fibo.py` no diretório atual com o seguinte conteúdo:

```python
# Módulo de números de Fibonacci

def fib_print(n):
    """Escreve a série de Fibonacci até n."""
    a, b = 0, 1
    while a < n:
        print(a, end=' ')
        a, b = b, a+b
    print()

def fib_build(n):
    """Retorna a série de Fibonacci até n."""
    result = []
    a, b = 0, 1
    while a < n:
        result.append(a)
        a, b = b, a+b
    return result
```

Agora, e importe esse módulo com o seguinte comando:

````python
import fibo
````

Isso adiciona o nome do módulo `fibo.py` no espaço de nomes do programa principal. Assim, usando o nome do módulo, você pode acessar as funções:

````python
fibo.fib_print(100)
````

````python
fibo.fib_build(100)
````

### 7.2. Mais sobre Módulos

Um módulo pode conter tanto instruções executáveis quanto definições de funções e classes. Essas instruções servem para inicializar o módulo. Eles são executados somente na *primeira* vez que o módulo é encontrado em uma instrução de importação. (Também rodam se o arquivo é executado como um *script*.)

Cada módulo tem seu próprio espaço de nomes privado, que é usado como espaço de nomes global para todas as funções definidas no módulo. Assim, o autor de um módulo pode usar variáveis globais no seu módulo sem se preocupar com conflitos acidentais com as variáveis globais do usuário. Por outro lado, se você precisar usar uma variável global de um módulo, poderá fazê-lo com a mesma notação usada para se referir às suas funções, `nomemodulo.nomeitem`.

Módulos podem importar outros módulos. É costume, porém não obrigatório, colocar todas as instruções import no início do módulo (ou script , se preferir). As definições do módulo importado, se colocados no nível de um módulo (fora de quaisquer funções ou classes), elas são adicionadas a espaço de nomes global da módulo.

Existe uma variante da instrução `import` que importa definições de um módulo diretamente para o espaço de nomes do módulo importador. Por exemplo:

````python
from fibo import fib_build, fib_print

fib_print(100)
fib_build(100)
````

Isso não coloca o nome do módulo de onde foram feitas as importações no espaço de nomes local (assim, no exemplo, `fibo` não está definido).

Existe ainda uma variante que importa todos os nomes definidos em um módulo:

````python
from fibo import *

fib_print(100)
````

Isso importa todos as declarações de nomes, exceto aqueles que iniciam com um sublinhado (`_`). Na maioria dos casos, programadores Python não usam esta facilidade porque ela introduz um conjunto desconhecido de nomes no ambiente, podendo esconder outros nomes previamente definidos.

Note que, em geral, a prática do `import *` de um módulo ou pacote é desaprovada, uma vez que muitas vezes dificulta a leitura do código. Contudo, é aceitável para diminuir a digitação em sessões interativas.

Se o nome do módulo é seguido pela palavra-chave `as`, o nome a seguir é vinculado diretamente ao módulo importado.

````python
import fibo as fib

fib.fib_print(100)
````

Isto efetivamente importa o módulo, da mesma maneira que `import fibo` fará, com a única diferença de estar disponível com o nome `fib`(como um *alias*).

Também pode ser utilizado com a palavra-chave `from`, com efeitos similares:

````python
from fibo import fib_print as fibonacci

fibonacci(100)
````

> Por motivos de eficiência, cada módulo é importado apenas uma vez por sessão do interpretador. Portanto, se você alterar seus módulos, deverá reiniciar o interpretador.

### 7.3. Pacotes

Os pacotes são uma maneira de estruturar o `"espaço de nomes"` dos módulos Python, usando `"nomes de módulo com pontos"`. Por exemplo, o nome do módulo `A.B` designa um submódulo chamado `B`, em um pacote chamado `A`. Assim como o uso de módulos evita que os autores de módulos diferentes tenham que se preocupar com nomes de variáveis globais, o uso de nomes de módulos com pontos evita que os autores de pacotes com muitos módulos, como `NumPy` ou `Pandas`, tenham que se preocupar com os nomes dos módulos uns dos outros.

Suponha que você queira projetar uma coleção de módulos (um `"pacote"`) para o gerenciamento uniforme de arquivos de som. Existem muitos formatos diferentes (normalmente identificados pela extensão do nome de arquivo, por exemplo `.wav`, `.aiff`, `.au`), de forma que você pode precisar criar e manter uma crescente coleção de módulos de conversão entre formatos. Ainda podem existir muitas operações diferentes, passíveis de aplicação sobre os arquivos de som (mixagem, eco, equalização, efeito stereo artificial). Logo, possivelmente você também estará escrevendo uma coleção sempre crescente de módulos para aplicar estas operações. Eis uma possível estrutura para o seu pacote (expressa em termos de um sistema de arquivos hierárquico):

```python
sound/                          Pacote de nível superior
      __init__.py               Inicializa o pacote de som
      formats/                  Subpacote para as conversões entre formatos de arquivos
              __init__.py
              wavread.py
              wavwrite.py
              aiffread.py
              aiffwrite.py
              auread.py
              auwrite.py
              ...
      effects/                  Subpacote para efeitos de som
              __init__.py
              echo.py
              surround.py
              reverse.py
              ...
      filters/                  Subpacote para filtros
              __init__.py
              equalizer.py
              vocoder.py
              karaoke.py
              ...
```

Os arquivos `__init__.py` são necessários para que o Python trate diretórios contendo o arquivo como pacotes (a menos que se esteja usando um pacote de espaço de nomes, um recurso relativamente avançado). Isso impede que diretórios com um nome comum, como string, ocultem, involuntariamente, módulos válidos que ocorrem posteriormente no caminho de busca do módulo. No caso mais simples, `__init__.py` pode ser apenas um arquivo vazio, mas pode também executar código de inicialização do pacote, ou configurar a variável `__all__`, descrita mais adiante.

Usuários do pacote podem importar módulos individuais, por exemplo:

```python
import sound.effects.echo
```

Isso carrega o submódulo `sound.effects.echo`. Ele deve ser referenciado com seu nome completo, como em:

```python
sound.effects.echo.echofilter(input, output, delay=0.7, atten=4)
```

Uma maneira alternativa para a importação desse módulo é:

```python
from sound.effects import echo
```

Isso carrega o submódulo echo sem necessidade de mencionar o prefixo do pacote no momento da utilização, assim:

```python
echo.echofilter(input, output, delay=0.7, atten=4)
```

Também é possível importar diretamente uma única variável ou função:

```python
from sound.effects.echo import echofilter
```

Novamente, isso carrega o submódulo echo, mas a função `echofilter()` está acessível diretamente sem prefixo:

```python
echofilter(input, output, delay=0.7, atten=4)
```

Observe que ao utilizar `from pacote import item`, o `item` pode ser um subpacote, submódulo, classe, função ou variável. A instrução `import` primeiro testa se o `item` está definido no pacote, senão presume que é um módulo e tenta carregá-lo. Se falhar em encontrar o módulo, uma exceção `ImportError` é levantada.

Em oposição, em uma construção como `import item.subitem.subsubitem`, cada item, com exceção do último, deve ser um pacote. O último pode ser também um pacote ou módulo, mas nunca uma classe, função ou variável contida em um módulo.

### 7.4. Importando * de um Pacote

Agora, o que acontece quando um usuário escreve `from sound.effects import *`? Idealmente, poderia se esperar que este comando vasculhasse o sistema de arquivos, encontrasse todos os submódulos presentes no pacote, e os importasse. Isso poderia demorar muito e a importação de submódulos pode ocasionar efeitos colaterais, que somente deveriam ocorrer quando o submódulo é explicitamente importado.

A única solução é o autor do pacote fornecer um índice explícito do pacote. A instrução `import` usa a seguinte convenção: se o arquivo `__init__.py` do pacote define uma lista chamada `__all__`, então esta lista indica os nomes dos módulos a serem importados quando a instrução `from pacote import *` é acionada. Fica a cargo do autor do pacote manter esta lista atualizada, inclusive fica a seu critério excluir inteiramente o suporte a importação direta de todo o pacote através de from pacote `import *`. Por exemplo, o arquivo `sounds/effects/__init__.py` poderia conter apenas:

```python
__all__ = ["echo", "surround", "reverse"]
```

Isso significaria que `from sound.effects import *` importaria os três submódulos nomeados do pacote `sound.effects`.

Esteja ciente de que os submódulos podem ficar sobrepostos por nomes definidos localmente. Por exemplo, se você adicionou uma função `reverse` ao arquivo `sound/effects/__init__.py`, usar `from sound.effects import *` só importaria os dois submódulos `echo e surround`, mas não o submódulo `reverse`, porque ele fica sobreposto pela função `reverse` definida localmente:

```python
__all__ = [
    "echo",      # refere-se ao arquivo 'echo.py'
    "surround",  # refere-se ao arquivo 'surround.py'
    "reverse",   # !!! refere-se à função 'reverse' agora !!!
]

def reverse(msg: str):  # <-- este nome ofusca o submódulo 'reverse.py'
    return msg[::-1]    #     no caso de uma importação 'from sound.effects import *'
```

Se `__all__` não estiver definido, a instrução `from sound.effects import *` não importa todos os submódulos do pacote `sound.effects` no espaço de nomes atual. Há apenas garantia que o pacote `sound.effects` foi importado (possivelmente executando qualquer código de inicialização em `__init__.py`) juntamente com os nomes definidos no pacote. Isso inclui todo nome definido em `__init__.py` bem como em qualquer submódulo importado a partir deste. Também inclui quaisquer submódulos do pacote que tenham sido carregados explicitamente por instruções `import` anteriores. Considere o código abaixo:

```python
import sound.effects.echo
import sound.effects.surround
from sound.effects import *
```

Nesse exemplo, os módulos `echo` e `surround` são importados no espaço de nomes atual, no momento em que a instrução `from...import` é executada, pois estão definidos no pacote `sound.effects`. (Isso também funciona quando `__all__` estiver definida.)

Apesar de que certos módulos são projetados para exportar apenas nomes conforme algum critério quando se faz `import *`, ainda assim essa sintaxe é considerada uma prática ruim em código de produção.

Lembre-se, não há nada errado em usar `from pacote import submodulo_especifico`! De fato, essa é a notação recomendada, a menos que o módulo importado necessite usar submódulos com o mesmo nome, de diferentes pacotes.

### 7.5. Biblioteca de Funções

Uma biblioteca de funções é um conjunto de rotinas ou procedimentos pré-definidos que podem ser chamados e reutilizados em um programa de computador. Um programa pode utilizar recursos de uma biblioteca a fim de contribuir para alcançar seu objetivo.

Em Python, podemos usar o conceito de módulos para criar bibliotécas. Um módulo pode conter quaisquer recursos e estruturas do Python: Variáveis, Funções e Classes. Por exemplo, considere a seguinte estrutura do pacote `matematica`:

```python
matematica/                          Pacote de nível superior
           __init__.py               Inicializa o pacote de matematica
           calculadora.py
```

O arquivo `__init__.py` contém o seguinte conteúdo:

```python
__all__ = [
    "calculadora"      # refere-se ao arquivo 'calculadora.py'
]
```

O arquivo `calculadora.py` contém o seguinte conteúdo:

```python
def sum(a, b):
    return a + b

def sub(a, b):
    return a - b

def mut(a, b):
    return a * b

def div(a, b):
    return a / b
```

O pacote `matematica` disponibiliza o módulo `calculadora` que funciona como uma bibliotéca de funções (uma vez que seu conteúdo e constituído apenas de funções). Assim, podemos reutilizar as funções definidas em `calculadora.py` através de um processo de importação da seguinte forma:

````python
from matematica import calculadora

print(calculadora.sum(10,12))
print(calculadora.sub(51,21))
print(calculadora.mut(11,10))
print(calculadora.div(20,10))
````

Podemos ainda dar uma **alias** para a bliblioteca a fim de melhorar a legibilidade ou evitar sobreposição de nomes de variáveis no programa:

````python
from matematica import calculadora as calc

print(calc.sum(10,12))
print(calc.sub(51,21))
print(calc.mut(11,10))
print(calc.div(20,10))
````

Podemos ainda fazer a importação de outra forma:

````python
from matematica.calculadora import *

print(sum(10,12))
print(sub(51,21))
print(mut(11,10))
print(div(20,10))
````

Nesse caso, fazemos o `import` de todas as funções de `calculadora`. Eventualmente, poderíamos importar apenas as funções desejadas:

````python
from matematica.calculadora import sum, sub

print(sum(10,12))
print(sub(51,21))
````

Observe que, nesse caso, se fizermos chamadas às demais funções, ocorrerá um erro do tipo `NameError`:

````python
print(mut(11,10))
print(div(20,10))
````

## 8. Orientação a Objetos

A programação orientada a objetos (POO) em Python ajuda você a estruturar seu código agrupando dados e comportamentos relacionados em **objetos**. Você começa definindo **classes**, que funcionam como modelos, e então cria objetos a partir delas. A POO simplifica a modelagem de conceitos do mundo real em seus programas e permite que você construa sistemas mais reutilizáveis ​​e escaláveis.

Nesta seção, exploraremos como definir classes, instanciar classes para criar objetos e aproveitar a herança para construir sistemas robustos em Python.

### 8.1. O que é Programação Orientada a Objetos?


A programação orientada a objetos é um paradigma de programação que fornece um meio de estruturar programas de forma que propriedades e comportamentos sejam agrupados em **objetos** individuais.

Por exemplo, um objeto pode representar uma pessoa com **propriedades** como nome, idade e endereço, e **comportamentos** como andar, falar, respirar e correr. Ou pode representar um e-mail com propriedades como lista de destinatários, assunto e corpo, e comportamentos como adicionar anexos e enviar.

Em outras palavras, a programação orientada a objetos é uma abordagem para modelar coisas concretas do mundo real, como carros, bem como relações entre coisas, como empresas e funcionários ou alunos e professores. A POO modela entidades do mundo real como objetos de software que possuem alguns dados associados a eles e podem executar determinadas operações.

A POO também existe em outras linguagens de programação e é frequentemente descrita como centrada nos quatro pilares, ou quatro princípios da POO:

- O **encapsulamento** permite agrupar dados (atributos) e comportamentos (métodos) dentro de uma classe para criar uma unidade coesa. Ao definir métodos para controlar o acesso a atributos e sua modificação, o encapsulamento ajuda a manter a integridade dos dados e promove um código modular e seguro.

- A **herança** permite a criação de relacionamentos hierárquicos entre classes, permitindo que uma subclasse herde atributos e métodos de uma classe pai. Isso promove a reutilização de código e reduz a duplicação.

- A **abstração** se concentra em ocultar detalhes de implementação e expor apenas a funcionalidade essencial de um objeto. Ao impor uma interface consistente, a abstração simplifica as interações com objetos, permitindo que os desenvolvedores se concentrem no que um objeto faz, em vez de como ele atinge sua funcionalidade.

- O **polimorfismo** permite tratar objetos de diferentes tipos como instâncias do mesmo tipo base, desde que implementem uma interface ou comportamento comum. A tipagem duck do Python o torna especialmente adequado para polimorfismo, pois permite acessar atributos e métodos em objetos sem precisar se preocupar com sua classe real.

Esta seção, você adotará uma abordagem prática para entender a POO em Python. Mas manter esses quatro conceitos de programação orientada a objetos em mente pode ajudá-lo a se lembrar das informações coletadas.

A principal conclusão é que os objetos estão no centro da programação orientada a objetos em Python. Em outros paradigmas de programação, os objetos representam apenas os dados. Em POO, eles também informam a estrutura geral do programa.

### 8.2. Como definir uma classe?

Em Python, você define uma classe usando a palavra-chave `class` seguida de um nome e dois pontos (`:`). Em seguida, você usa `.__init__()` para declarar quais atributos cada instância da classe deve ter:

````python
class Employee:
    def __init__(self, name, age):
        self.name =  name
        self.age = age
````

Mas o que tudo isso significa? E por que você precisa de classes, para começo de conversa? Dê um passo para trás e considere usar estruturas de dados primitivas integradas como alternativa.

Estruturas de dados primitivas — como números, strings e listas — são projetadas para representar informações simples, como o preço de uma maçã, o nome de um poema ou suas cores favoritas, respectivamente. E se você quiser representar algo mais complexo?

Por exemplo, você pode querer rastrear funcionários em uma organização. Você precisa armazenar algumas informações básicas sobre cada funcionário, como nome, idade, cargo e ano em que começou a trabalhar.

Uma maneira de fazer isso é representar cada funcionário como uma lista:

````python
kirk = ["James Kirk", 34, "Captain", 2265]
spock = ["Spock", 35, "Science Officer", 2254]
mccoy = ["Leonard McCoy", "Chief Medical Officer", 2266]
````

Há vários problemas com essa abordagem.

Primeiro, ela pode dificultar o gerenciamento de arquivos de código maiores. Se você referenciar `kirk` várias linhas depois de onde declarou a lista `kirk`, lembrará que o elemento com índice `0` é o `nome` do funcionário?

Segundo, ela pode introduzir erros se os funcionários não tiverem o mesmo número de elementos em suas respectivas listas. Na lista `mccoy` acima, a idade está ausente, então `mccoy` retornará `"Chief Medical Officer"` em vez da idade do Dr. McCoy.

Uma ótima maneira de tornar esse tipo de código mais gerenciável e de fácil manutenção é usar classes.

### 8.3. Classes vs Instâncias

Classes permitem criar estruturas de dados definidas pelo usuário. Classes definem funções especiais chamadas **métodos**, que identificam os comportamentos e ações que um objeto criado a partir da classe pode executar com seus dados.

Nesta seção, usaremos uma classe `Dog` que armazena algumas informações sobre as características e comportamentos que um cachorro pode ter.

Uma classe é um modelo de como definir algo. Na verdade, ela não contém nenhum dado. A classe `Dog` especifica que um nome e uma idade são necessários para definir um cachorro, mas não contém o nome ou a idade de nenhum cachorro específico.

Enquanto a classe é o modelo, uma **instância** é um objeto construído a partir de uma classe e contém dados reais. Uma instância da classe `Dog` não é mais um modelo. É um cachorro de verdade com um nome, como `'Miles'`, que tem quatro anos.

Em outras palavras, uma classe é como um formulário ou questionário. Uma instância é como um formulário que você preencheu com informações. Assim como muitas pessoas podem preencher o mesmo formulário com suas próprias informações exclusivas, você pode criar muitas instâncias de uma única classe.

### 8.4. Definição de Classe

Inicie todas as definições de classe com a palavra-chave `class` e, em seguida, adicione o nome da classe e dois pontos (`:`). O Python considerará qualquer código que você recuar abaixo da definição da classe como parte do corpo da classe.

Aqui está um exemplo da classe `Dog`:

````python
class Dog:
    pass
````

O corpo da classe `Dog` consiste em uma única instrução: a palavra-chave `pass`. Programadores Python costumam usar pass como um espaço reservado para indicar onde o código será inserido. Isso permite que você execute esse código sem que o Python gere um erro.

A classe `Dog` não é muito interessante agora, vamos incrementá-la um pouco definindo algumas propriedades que todos os objetos `Dog` devem ter. Há várias propriedades que você pode escolher, incluindo nome, idade, cor da pelagem e raça. Para manter o escopo do exemplo pequeno, usaremos apenas nome e idade.

Definimos as propriedades que todos os objetos `Dog` devem ter em um método chamado `.__init__()`. Toda vez que você cria um novo objeto `Dog`, `.__init__()` define o **estado inicial** do objeto atribuindo os valores das propriedades do objeto. Ou seja, `.__init__()` inicializa cada nova instância da classe.

Podemos atribuir a `.__init__()` qualquer número de parâmetros, mas o primeiro parâmetro sempre será uma variável chamada `self`. Ao criar uma nova instância de classe, o Python passa automaticamente a instância para o parâmetro `self` em `.__init__()` para que o Python possa definir os novos atributos no objeto.

Vamos atualizar a classe `Dog` com um método `.__init__()` que cria atributos `.name` e `.age`:

````python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age
````

Precisamos nos certificar de recuar a assinatura do método `.__init__()` em quatro espaços e o corpo do método em oito espaços. Esse recuo é de vital importância. Ele informa ao Python que o método .`__init__()` pertence à classe `Dog`.

No corpo de `.__init__()`, há duas instruções usando a variável `self`:

- `self.name = name` cria um atributo chamado `name` e atribui a ele o valor do parâmetro `name`.
- `self.age = age` cria um atributo chamado `age` e atribui a ele o valor do parâmetro `age`.

Atributos criados em `.__init__()` são chamados de atributos de instância. O valor de um atributo de instância é específico para uma instância específica da classe. Todos os objetos `Dog` têm um nome e uma idade, mas os valores dos atributos `name` e `age` variam dependendo da instância `Dog`.

Por outro lado, atributos de classe são atributos que têm o mesmo valor para todas as instâncias de classe. Você pode definir um atributo de classe atribuindo um valor a um nome de variável fora de `.__init__()`.

Por exemplo, a classe `Dog` a seguir tem um atributo de classe chamado `species` com o valor `"Canis familiaris"`:

````python
class Dog:
    species = "Canis familiaris"

    def __init__(self, name, age):
        self.name = name
        self.age = age
````

Defina os atributos de classe diretamente abaixo da primeira linha do nome da classe e os indentie com quatro espaços. Você sempre precisa atribuir um valor inicial a eles. Ao criar uma instância da classe, o Python cria e atribui automaticamente atributos de classe aos seus valores iniciais.

Use atributos de classe para definir propriedades que devem ter o mesmo valor para todas as instâncias da classe. Use atributos de instância para propriedades que variam de uma instância para outra.

Agora que você temos uma classe `Dog`, é hora de criar alguns cachorros!

### 8.5. Criando Instâncias

Criar um novo objeto a partir de uma classe é chamado de **instanciação de classe**. Você pode criar um novo objeto digitando o nome da classe, seguido de abrir e fechar parênteses:

````python
class Dog:
    pass

Dog()
````

Primeiro, criamos uma nova classe `Dog` sem atributos ou métodos e, em seguida, instanciamos a classe `Dog` para criar um objeto.

Na saída acima, você pode ver que agora tem um novo objeto Dog em `0x106d76c50`. Essa sequência estranha de letras e números é um endereço de memória que indica onde o Python armazena o objeto da classe Dog na memória do seu computador. Observe que o endereço na tela será diferente.

Agora, ao instancia a classe Dog uma segunda vez para criar outro objeto:

````python
Dog()
````

A nova instância de `Dog` está localizada em um endereço de memória diferente. Isso ocorre porque é uma instância inteiramente nova e completamente diferente do primeiro objeto `Dog` que foi criado.

Para ver isso de outra forma:

````python
a = Dog()
b = Dog()
a == b
````

Neste código, você cria dois novos objetos `Dog` e os atribui às variáveis `​a` e `b`. Ao comparar `a` e `b` usando o operador `==`, o resultado é `False`. Embora `a` e `b` sejam instâncias da classe `Dog`, eles representam dois objetos distintos na memória.

### 8.6. Classes e Atributos de Instância

Criando uma nova classe `Dog` com um atributo de classe chamado `.species` e dois atributos de instância chamados `.name` e `.age`:

````python
class Dog:
    species = "Canis familiaris"

    def __init__(self, name, age):
        self.name = name
        self.age = age
````

Para instanciar esta classe `Dog`, precisamos fornecer valores para `name` e `age`. Caso contrário, o Python gerará um `TypeError`:

````python
Dog()
````

Para passar argumentos para os parâmetros `name` e `age`, devemos colocar os valores entre parênteses após o nome da classe:

````python
miles = Dog("Miles", 4)
buddy = Dog("Buddy", 9)
````

Isso cria duas novas instâncias de `Dog` — uma para um cachorro de quatro anos chamado `Miles` e outra para um cachorro de nove anos chamado `Buddy`.

O método `.__init__()` da classe Dog tem três parâmetros, então por que você está passando apenas dois argumentos para ele no exemplo?

Ao instanciar a classe `Dog`, o Python cria uma nova instância de `Dog` e a passa para o primeiro parâmetro de `.__init__()`. Isso essencialmente remove o parâmetro `self`, então você só precisa se preocupar com os parâmetros `name` e `age`.

Depois de criar as instâncias de `Dog`, podemos acessar seus atributos de instância usando a notação de ponto:

````python
print(miles.name, miles.age)
print(buddy.name, buddy.age)
````

Podemos acessar atributos de classe da mesma maneira:

````python
print(buddy.species)
````

Uma das maiores vantagens de usar classes para organizar dados é que as instâncias têm a garantia de ter os atributos que você espera. Todas as instâncias de `Dog` têm os atributos `.species`, `.name` e `.age`, então você pode usar esses atributos com confiança, sabendo que eles sempre retornarão um valor.

Embora seja garantido que os atributos existam, seus valores podem mudar dinamicamente:

````python
buddy.age = 10
print(buddy.age)
miles.species = "Felis silvestris"
print(miles.species)
````

Neste exemplo, alteramos o atributo `.age` do objeto `buddy` para `10`. Em seguida, alteramos o atributo `.species` do objeto `miles` para `"Felis silvestris"`, que é uma espécie de gato. Isso torna `Miles` um cachorro bem estranho, mas é um Python válido!

A principal lição aqui é que objetos personalizados são mutáveis ​​por padrão. Um objeto é mutável se você puder alterá-lo dinamicamente. Por exemplo, listas e dicionários são mutáveis, mas strings e tuplas são imutáveis.

### 8.7. Métodos de Instância

Métodos de instância são funções que você define dentro de uma classe e só podem ser chamadas em uma instância dessa classe. Assim como `.__init__()`, um método de instância sempre recebe `self` como seu primeiro parâmetro.

````python
class Dog:
    species = "Canis familiaris"

    def __init__(self, name, age):
        self.name = name
        self.age = age

    # Método de instância
    def description(self):
        return f"{self.name} is {self.age} years old"

    # Outro método de instância
    def speak(self, sound):
        return f"{self.name} says {sound}"
````

Esta classe `Dog` possui dois métodos de instância:

- `.description()` retorna uma string exibindo o nome e a idade do cachorro.

- `.speak()` possui um parâmetro chamado `sound` e retorna uma string contendo o nome do cachorro e o som que ele emite.

Vamos ver os métodos de instância em ação:

````python
miles = Dog("Miles", 4)
print(miles.description())
print(miles.speak("Woof Woof"))
print(miles.speak("Bow Wow"))
````

Na classe `Dog` acima, `.description()` retorna uma string contendo informações sobre a instância de `Dog`, `miles`. Ao escrever suas próprias classes, é uma boa ideia ter um método que retorne uma string contendo informações úteis sobre uma instância da classe. No entanto, `.description()` não é a maneira mais Pythonica de fazer isso.

Ao criar um objeto de lista, você pode usar `print()` para exibir uma string que se parece com a lista:

````python
names = ["Miles", "Buddy", "Jack"]
print(names)
````

Vamos em frente tentar imprir o objeto `miles` para ver qual saída obtemos:

````python
print(miles)
````

Ao imprimir `miles`, você recebe uma mensagem aparentemente enigmática informando que `miles` é um objeto `Dog` no endereço de memória `0x106d12bd0`. Esta mensagem não é muito útil. Você pode alterar o que é impresso definindo um método de instância especial chamado `.__str__()`.

Vamos alterar o nome do método `.description()` da classe `Dog` para `.__str__()`:

````python
class Dog:
    species = "Canis familiaris"

    def __init__(self, name, age):
        self.name = name
        self.age = age

    # Instance method
    def __str__(self):
        return f"{self.name} is {self.age} years old"

    # Another instance method
    def speak(self, sound):
        return f"{self.name} says {sound}"
````

Agora, vejamos o que acontece quando tentamos imprimir `miles`:

````python
miles = Dog("Miles", 4)
print(miles)
````

Métodos como `.__init__()` e `.__str__()` são chamados de métodos `"dunder"` porque começam e terminam com dois sublinhados. Existem muitos métodos `"dunder"` que você pode usar para personalizar classes em Python. Entender os métodos `"dunder"` é uma parte importante do domínio da programação orientada a objetos em Python.

### 8.8. Herança

Herança é o processo pelo qual uma classe assume os atributos e métodos de outra. Classes recém-formadas são chamadas de **classes filhas**, e as classes das quais você deriva classes filhas são chamadas de **classes pais**.

Você herda de uma classe pai criando uma nova classe e colocando o nome da classe pai entre parênteses:

````python
class Animal:

    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def __str__(self):
        return f"{self.name} is {self.age} years old"

class Dog(Animal):
    def __init__(self, name, age, race):
        super().__init__(name, age)
        self.race = race
    
    def speak(self, sound):
        return f"{self.name} says {sound}"
````

Nesse exemplo, a classe `Dog` (filha) herda da classe `Animal` (pai). `Dog` herda os atributos `.name` e `.age` e o método `.__str__()` de `Animal`. Observe que no contrutor de `Dog`, o método especial `supper()` é invovado. Isso permite fazer uma refer6encia à classe pai e, no caso em questão, invocar o contrutor da classe pai passando como argumento o nome e a idade do cachorro (`super()__init__(name, age)`).

Agora, vamos explorar o resultado:

````python
miles = Dog("Miles", 4, "Labrador")
print(miles)
print(miles.speak("Woof Woof"))
print(miles.race)
````
