### 1.1. Visão Geral

Apresentaremos uma visão geral da mecânica dessas funções, que visam converter dados de texto em um `DataFrame`. Os argumentos opcionais para essas funções podem se enquadrar em algumas categorias:

- *Indexação*. Pode tratar uma ou mais colunas como o `DataFrame` retornado e determinar se deseja obter
nomes de colunas do arquivo, argumentos fornecidos por você ou não.

- *Inferência de tipos e conversão de dados*. Inclui as conversões de valores definidas pelo usuário e uma lista personalizada de marcadores de valores ausentes.

- *Análise de data e hora*. Inclui um recurso de combinação, incluindo a combinação de informações de data e hora distribuídas por várias colunas em uma única coluna no resultado.

- *Iteração*. Suporte para iteração em partes de arquivos muito grandes.

- *Problemas com dados sujos*. Inclui pular linhas ou um rodapé, comentários ou outras coisas menores, como dados numéricos com milhares separados por vírgulas.

Devido à confusão de dados no mundo real, algumas das funções de carregamento de dados (especialmente `pandas.read_csv`) acumularam uma longa lista de argumentos opcionais ao longo do tempo. É normal se sentir sobrecarregado com o número de parâmetros diferentes (`pandas.read_csv` tem cerca de 50). A [documentação online do Pandas](https://pandas.pydata.org/docs/) tem muitos exemplos sobre como cada um deles funciona, então se você estiver com dificuldades para ler um arquivo específico, pode haver um exemplo semelhante o suficiente para ajudá-lo a encontrar os parâmetros corretos.

Algumas dessas funções realizam inferência de tipos, pois os tipos de dados das colunas não fazem parte do formato dos dados. Isso significa que você não precisa necessariamente especificar quais colunas são numéricas, inteiras, booleanas ou strings. Outros formatos de dados, como `HDF5`, `ORC` e `​​Parquet`, possuem as informações do tipo de dados incorporadas ao formato.

Ao processar arquivos muito grandes ou descobrir o conjunto correto de argumentos para processar corretamente um arquivo grande, você pode querer ler apenas uma pequena parte do arquivo ou iterar por partes menores.

Antes de analisarmos um arquivo grande, tornamos as configurações de exibição do Pandas mais compactas:

Para ler um arquivo em partes, especifique um `chunksize` como um número de linhas:

````python
chunker = pd.read_csv("examples/ex6.csv", chunksize=1000)
````

````python
type(chunker)
````

O objeto `TextFileReader` retornado por `pandas.read_csv` permite iterar sobre as partes do arquivo de acordo com o `chunksize`. Por exemplo, podemos iterar sobre `ex6.csv`, agregando as contagens de valores na coluna "key", assim:

````python
chunker = pd.read_csv("examples/ex6.csv", chunksize=1000)
tot = pd.Series([], dtype='int64')
for piece in chunker:
    tot = tot.add(piece["key"].value_counts(), fill_value=0)

tot = tot.sort_values(ascending=False)
tot[:10]
````

`TextFileReader` também é equipado com um método `get_chunk` que permite ler pedaços de tamanho arbitrário.

### 1.4. Trabalhando com Outros Formatos Delimitados

É possível carregar a maioria dos formatos de dados tabulares do disco usando funções como `pandas.read_csv`. Em alguns casos, no entanto, pode ser necessário algum processamento manual. Não é incomum receber um arquivo com uma ou mais linhas malformadas que fazem com que o `pandas.read_csv` falhe. Para ilustrar as ferramentas básicas, considere um pequeno arquivo CSV:

````python
!cat examples/ex7.csv
````

Para qualquer arquivo com um delimitador de caractere único, você pode usar o módulo `csv` integrado do Python. Para usá-lo, passe qualquer arquivo aberto ou objeto semelhante a um arquivo para `csv.reader`:

````python
import csv

file = open("/home/samuelmalves/Documents/Dev/Python/CTE-IA/prog-ciencia-de-dados/aulas/examples/ex7.csv")
reader = csv.reader(file)
````

A iteração pelo leitor como um arquivo produz listas de valores com quaisquer caracteres de aspas removidos:

````python
for line in reader:
    print(line)
````

A partir daí, cabe a você fazer o trabalho necessário para colocar os dados no formato desejado. Vamos passo a passo. Primeiro, lemos o arquivo em uma lista de linhas:

````python
with open("/home/samuelmalves/Documents/Dev/Python/CTE-IA/prog-ciencia-de-dados/aulas/examples/ex7.csv") as file:
    lines = list(csv.reader(file))
````

Em seguida, dividimos as linhas em linha de cabeçalho e linhas de dados:

````python
header, values = lines[0], lines[1:]
````

Então podemos criar um dicionário de colunas de dados usando uma compreensão de dicionário e a expressão `zip(*values)` (cuidado, pois isso usará muita memória em arquivos grandes), que transpõe linhas para colunas:

````python
data_dict = {h: v for h, v in zip(header, zip(*values))}
data_dict
````

Os arquivos CSV estão disponíveis em diversos formatos. Para definir um novo formato com um delimitador, convenção de aspas de string ou terminador de linha diferente, poderíamos definir uma subclasse simples de `csv.Dialect`:

````python
file = open("/home/samuelmalves/Documents/Dev/Python/CTE-IA/prog-ciencia-de-dados/aulas/examples/ex7.csv")

class MyDialect(csv.Dialect):
    lineterminator = "\n"
    delimiter = ";"
    quotechar = '"'
    quoting = csv.QUOTE_MINIMAL

reader = csv.reader(file, dialect=MyDialect)
````

Também poderíamos fornecer parâmetros de dialeto CSV individuais como palavras-chave para `csv.reader` sem precisar definir uma subclasse:

````python
reader = csv.reader(file, delimiter="|")
````

As opções possíveis (atributos de csv.Dialect) e o que elas fazem podem ser encontradas na tabela abaixo

| Argumento | Descrição |
|------------|------------|
| `delimiter` | Cadeia de caracteres de um único caractere usada para separar campos; o padrão é `","`. |
| `lineterminator` | Terminador de linha para escrita; o padrão é `"\r\n"`. O leitor ignora isso e reconhece terminadores de linha compatíveis entre plataformas. |
| `quotechar` | Caractere de aspas usado para campos com caracteres especiais (como o delimitador); o padrão é `"'"`. |
| `quoting` | Convenção de aspas. As opções incluem: `csv.QUOTE_ALL` (aspas em todos os campos), `csv.QUOTE_MINIMAL` (somente em campos com caracteres especiais, como o delimitador), `csv.QUOTE_NONNUMERIC` e `csv.QUOTE_NONE` (sem aspas). O padrão é `QUOTE_MINIMAL`. |
| `skipinitialspace` | Ignora espaços em branco após cada delimitador; o padrão é `False`. |
| `doublequote` | Define como lidar com caracteres de aspas dentro de um campo; se `True`, o caractere de aspas é duplicado (veja a documentação online para mais detalhes). |
| `escapechar` | Caractere usado para escapar o delimitador se `quoting` estiver definido como `csv.QUOTE_NONE`; desativado por padrão. |

Para gravar arquivos delimitados manualmente, você pode usar csv.writer. Ele aceita um objeto de arquivo aberto e gravável e as mesmas opções de dialeto e formato que `csv.reader`:

````python
with open("mydata.csv", "w") as f:
    writer = csv.writer(f, dialect=MyDialect)
    writer.writerow(("one", "two", "three"))
    writer.writerow(("1", "2", "3"))
    writer.writerow(("4", "5", "6"))
    writer.writerow(("7", "8", "9"))
````

### 1.5. Dados no Formato JSON

JSON (abreviação de *JavaScript Object Notation*) tornou-se um dos formatos padrão para envio de dados por solicitação HTTP entre navegadores da web e outros aplicativos. É um formato de dados muito mais livre do que um formato de texto tabular como CSV. Aqui está um exemplo:

````python
obj ="""
{"name": "Wes",
"pet": null,
"cities_lived": ["Akron", "Nashville", "New York", "San Francisco"],
"siblings": [{"name": "Scott", "age": 34, "hobbies": ["guitars", "soccer"]},
{"name": "Katie", "age": 42, "hobbies": ["diving", "art"]}]
}
"""
````

JSON é um código Python quase válido, com exceção de seu valor nulo, `null`, e algumas outras nuances (como a proibição de vírgulas finais no final de listas). Os tipos básicos são objetos (dicionários), arrays (listas), strings, números, booleanos e nulos. Todas as chaves em um objeto devem ser strings. Existem várias bibliotecas Python para ler e escrever dados JSON. Usaremos `json`, pois ela é integrada à biblioteca padrão do Python. Para converter uma string JSON para o formato Python, use `json.loads`:

````python
import json
result = json.loads(obj)
result
````

`json.dumps`, por outro lado, converte um objeto Python de volta para JSON:

````python
asjson = json.dumps(result)
asjson
````

A maneira de converter um objeto JSON ou uma lista de objetos em um `DataFrame` ou alguma outra estrutura de dados para análise dependerá de você. Convenientemente, você pode passar uma lista de dicionários (que anteriormente eram objetos JSON) para o construtor do `DataFrame` e selecionar um subconjunto dos campos de dados:

````python
siblings = pd.DataFrame(result["siblings"], columns=["name", "age"])
siblings
````

O `pandas.read_json` pode converter automaticamente conjuntos de dados JSON em arranjos específicos em uma `Series` ou `DataFrame`. Por exemplo:

````python
!cat examples/example.json
````

XML (*Extensible Markup Language*) é outro formato comum de dados estruturados que suporta dados hierárquicos e aninhados com metadados.

Vimos que a função `pandas.read_html` usa `lxml` ou `Beautiful Soup` para analisar dados de HTML. XML e HTML são estruturalmente semelhantes, mas XML é mais genérico. Assim, mostraremos um exemplo de como usar `lxml` para analisar dados de um formato XML mais genérico.

Por muitos anos, a Autoridade de Transporte Metropolitano de Nova York (MTA) publicou uma série de dados sobre seus serviços de ônibus e trens em formato XML. Aqui, veremos os dados de desempenho, que estão contidos em um conjunto de arquivos XML. Cada serviço de trem ou ônibus tem um arquivo diferente (como *Performance_MNR.xml* para a Ferrovia Metro-North) contendo dados mensais como uma série de registros XML semelhantes a este:

```xml
<INDICATOR>
<INDICATOR_SEQ>373889</INDICATOR_SEQ>
<PARENT_SEQ></PARENT_SEQ>
<AGENCY_NAME>Metro-North Railroad</AGENCY_NAME>
<INDICATOR_NAME>Escalator Availability</INDICATOR_NAME>
<DESCRIPTION>Percent of the time that escalators are operational
systemwide. The availability rate is based on physical observations performed
the morning of regular business days only. This is a new indicator the agency
began reporting in 2009.</DESCRIPTION>
<PERIOD_YEAR>2011</PERIOD_YEAR>
<PERIOD_MONTH>12</PERIOD_MONTH>
<CATEGORY>Service Indicators</CATEGORY>
<FREQUENCY>M</FREQUENCY>
<DESIRED_CHANGE>U</DESIRED_CHANGE>
<INDICATOR_UNIT>%</INDICATOR_UNIT>
<DECIMAL_PLACES>1</DECIMAL_PLACES>
<YTD_TARGET>97.00</YTD_TARGET>
<YTD_ACTUAL></YTD_ACTUAL>
<MONTHLY_TARGET>97.00</MONTHLY_TARGET>
<MONTHLY_ACTUAL></MONTHLY_ACTUAL>
</INDICATOR>
```

Usando `lxml.objectify`, analisamos o arquivo e obtemos uma referência ao nó raiz do arquivo XML com `getroot`:

````python
from lxml import objectify

path = "/home/samuelmalves/Documents/Dev/Python/CTE-IA/prog-ciencia-de-dados/aulas/datasets/mta_perf/Performance_MNR.xml"

with open(path) as f:
    parsed = objectify.parse(f)

root = parsed.getroot()
````

`root.INDICATOR` retorna um gerador que gera cada elemento XML `<INDICATOR>`. Para cada registro, podemos preencher um dicionário de nomes de tags (como `YTD_ACTUAL`) com valores de dados (excluindo algumas tags) executando o seguinte código:

````python
data = []
skip_fields = ["PARENT_SEQ", "INDICATOR_SEQ",
               "DESIRED_CHANGE", "DECIMAL_PLACES"]
for elt in root.INDICATOR:
    el_data = {}
    for child in elt.getchildren():
        if child.tag in skip_fields:
            continue
        el_data[child.tag] = child.pyval
    data.append(el_data)
````

Por fim, converta esta lista de dicionários em um `DataFrame`:

````python
perf = pd.DataFrame(data)
perf.head()
````

A função `pandas.read_xml` do Pandas transforma esse processo em uma expressão de uma linha:

````python
perf_2 = pd.read_xml(path)
perf_2.head()
````

Para documentos XML mais complexos, consulte a docstring para [pandas.read_xml](https://pandas.pydata.org/docs/reference/api/pandas.read_xml.html), que descreve como fazer seleções e filtros para extrair uma tabela específica de interesse.

### 2.3. Usando o Formato HDF5

HDF5 (*Hierarchical Data Format version 5*) é um formato de arquivo respeitado, destinado ao armazenamento de grandes quantidades de dados científicos em arrays. Está disponível como uma biblioteca `C` e possui interfaces disponíveis em muitas outras linguagens, incluindo `Java`, `Julia`, `MATLAB` e `Python`. O `"HDF"` em `HDF5` significa formato de dados hierárquico. Cada arquivo HDF5 pode armazenar vários conjuntos de dados e metadados de suporte. Comparado a formatos mais simples, o HDF5 suporta compactação instantânea com uma variedade de modos de compactação, permitindo que dados com padrões repetidos sejam armazenados com mais eficiência. O HDF5 pode ser uma boa opção para trabalhar com conjuntos de dados que não cabem na memória, pois você pode ler e gravar com eficiência pequenas seções de arrays muito maiores.

Para começar a usar o HDF5 e o pandas, você deve primeiro instalar o `PyTables` instalando o pacote `tables` com `pip`:

````python
pip install tables
````

Embora seja possível acessar arquivos HDF5 diretamente usando as bibliotecas `PyTables` ou `h5py`, o Pandas oferece uma interface de alto nível que simplifica o armazenamento de objetos `Series` e `DataFrame`. A classe `HDFStore` funciona como um dicionário e lida com os detalhes de baixo nível:

````python
import numpy as np
import pandas as pd

frame = pd.DataFrame({"a": np.random.standard_normal(100)})
store = pd.HDFStore("/home/samuelmalves/Documents/Dev/Python/CTE-IA/prog-ciencia-de-dados/aulas/examples/mydata.h5")
store["obj1"] = frame
store["obj1_col"] = frame["a"]
store
````

Os objetos contidos no arquivo HDF5 podem então ser recuperados com a mesma API semelhante a um dicionário:

````python
store["obj1"]
````

O HDFStore suporta dois esquemas de armazenamento: `"fixed"` e `"table"` (o padrão é `"fixed"`). Este último é geralmente mais lento, mas suporta operações de consulta usando uma sintaxe especial:

````python
store.put("obj2", frame, format="table")
````

````python
store.select("obj2", where=["index >= 10 and index <= 15"])
````

````python
store.close()
````

O put é uma versão explícita do método `store["obj2"] = frame`, mas nos permite definir outras opções, como o formato de armazenamento.

A função `pandas.read_hdf` fornece um atalho para estas ferramentas:

````python
frame.to_hdf(path_or_buf="/home/samuelmalves/Documents/Dev/Python/CTE-IA/prog-ciencia-de-dados/aulas/examples/mydata.h5", key="obj3", format="table")
````

````python
pd.read_hdf("/home/samuelmalves/Documents/Dev/Python/CTE-IA/prog-ciencia-de-dados/aulas/examples/mydata.h5", "obj3", where=["index < 5"])
````

## 4. Interagindo com Bancos de Dados

Em um ambiente empresarial, muitos dados podem não ser armazenados em arquivos de texto ou Excel. Bancos de dados relacionais baseados em SQL (como SQL Server, PostgreSQL e MySQL) são amplamente utilizados, e muitos bancos de dados alternativos se tornaram bastante populares. A escolha do banco de dados geralmente depende das necessidades de desempenho, integridade dos dados e escalabilidade de uma aplicação.

O Pandas possui algumas funções para simplificar o carregamento dos resultados de uma consulta SQL em um `DataFrame`. Como exemplo, criaremos um banco de dados `SQLite3` usando o driver `sqlite3` integrado do Python:

````python
import sqlite3

query ="""
CREATE TABLE test
(a VARCHAR(20), b VARCHAR(20),
c REAL, d INTEGER
);"""
````

````python
con = sqlite3.connect("mydata.sqlite")
````

````python
con.execute(query)
````

````python
con.commit()
````

Em seguida, insira algumas linhas de dados:

````python
data = [("Atlanta", "Georgia", 1.25, 6), 
        ("Tallahassee", "Florida", 2.6, 3), 
        ("Sacramento", "California", 1.7, 5)]

stmt = "INSERT INTO test VALUES(?, ?, ?, ?)"

con.executemany(stmt, data)
````

````python
con.commit()
````

A maioria dos drivers Python SQL retorna uma lista de tuplas ao selecionar dados de uma tabela:

````python
cursor = con.execute("SELECT * FROM test")
````

````python
rows = cursor.fetchall()
rows
````

Você pode passar a lista de tuplas para o construtor `DataFrame`, mas também precisa dos nomes das colunas, contidos no atributo description do cursor. Observe que, para o `SQLite3`, a descrição do cursor fornece apenas os nomes das colunas (os outros campos, que fazem parte da especificação da API de Banco de Dados do Python, são `None`), mas para alguns outros drivers de banco de dados, mais informações sobre as colunas são fornecidas:

````python
cursor.description
````

````python
pd.DataFrame(rows, columns=[x[0] for x in cursor.description])
````

Está é uma `"confusão"` que você vai preferir não repetir sempre que consultar o banco de dados. O projeto `SQLAlchemy` é um popular kit de ferramentas SQL em Python que abstrai muitas das diferenças comuns entre bancos de dados SQL. O Pandas tem uma função `read_sql` que permite ler dados facilmente de uma conexão `SQLAlchemy` geral. Você pode instalar o `SQLAlchemy` com `pip` assim:

````python
pip install sqlalchemy
````

````python
import sqlalchemy as sqla
````

````python
db = sqla.create_engine("sqlite:///mydata.sqlite")
````

````python
pd.read_sql("SELECT * FROM test", db)
````
